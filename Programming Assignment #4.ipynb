```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
b = pd.read_excel('board2.xlsx')
b
```

```python
#Calculating the Average
subj_column = ["Math", "Electronics", "GEAS", "Communication"]
b["Average"] = b[subj_column].mean(axis=1)
```

**A. VISAYAS COMMUNICATION DATAFRAME**

```python
vis_comm_mask = (b["Hometown"] == "Visayas") & (b["Track"] == "Communication")
```

```python
VisComm = b.loc[vis_comm_mask, ["Name", "Gender", "Math", "Electronics", "Average"]]
VisComm
```

**B. VISAYAS FEMALE DATAFRAME**

```python
vis_female_mask = (b["Hometown"] == "Visayas") & (b["Gender"] == "Female")
```

```python
VisFemale = b.loc[vis_female_mask, ["Name", "Track", "GEAS", "Electronics", "Average"]]
VisFemale
```

```python
vis_fem_passed = VisFemale.loc[VisFemale["Average"] >= 60]
vis_fem_passed
```

**C. CATEGORY-AVERAGE VISUALIZATION**

```python
track_mean = b.groupby("Track")["Average"].mean()
track_mean
```

```python
gender_mean = b.groupby("Gender")["Average"].mean()
gender_mean
```

```python
hometown_mean = b.groupby("Hometown")["Average"].mean()
hometown_mean
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharey=True)

axes[0].bar(track_mean.index, track_mean.values, color="red", edgecolor="black")
axes[0].set_title("Mean Average by Track")
axes[0].set_xlabel("Track")
axes[0].set_ylabel("Mean Average")
axes[0].tick_params(axis="x", rotation=45)

axes[1].bar(gender_mean.index, gender_mean.values, color="orange", edgecolor="black")
axes[1].set_title("Mean Average by Gender")
axes[1].set_xlabel("Gender")
axes[1].tick_params(axis="x", rotation=0)

axes[2].bar(hometown_mean.index, hometown_mean.values, color="yellow",edgecolor="black")

axes[2].set_title("Mean Average by Hometown")
axes[2].set_xlabel("Hometown")
axes[2].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

```python
print("\n**The Highest Sample Mean Average of Each Category**")
print(f"1. Track: {track_mean.idxmax()} ({track_mean.max():.2f}).")
print(f"2. Gender: {gender_mean.idxmax()} ({gender_mean.max():.2f}).")
print(f"3. Hometown: {hometown_mean.idxmax()} ({hometown_mean.max():.2f}).")
```
